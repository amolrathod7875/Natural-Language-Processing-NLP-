```python
import nltk
from nltk.sentiment import SentimentIntensityAnalyzer
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
```

```python
# Download VADER lexicon
nltk.download('vader_lexicon', quiet=True)
```

## Sample Twitter Data

```python
tweets = [
    "I absolutely LOVE the new iPhone! Best purchase ever! 😍🔥",
    "This airline lost my luggage again. Worst service EVER. 😡",
    "Just had coffee and a bagel. Pretty standard morning.",
    "The new Marvel movie was incredible! Mind-blowing visuals! 🎬",
    "I'm so frustrated with this software. Keeps crashing nonstop. 😤",
    "Looking forward to the weekend. Gonna relax and enjoy! 🌟",
    "Traffic was terrible this morning. Late again. Ugh. 🚗",
    "Amazing customer support from @Amazon. They resolved my issue in minutes!",
    "Meh, the food was okay. Nothing special tbh.",
    "Just won a free gift card! Today is a GREAT day! 🎉"
]
```

## Analyze Sentiment Using VADER

```python
sia = SentimentIntensityAnalyzer()

results = []
for tweet in tweets:
    scores = sia.polarity_scores(tweet)
    compound = scores['compound']

    if compound >= 0.05:
        label = "Positive"
    elif compound <= -0.05:
        label = "Negative"
    else:
        label = "Neutral"

    results.append({
        "tweet": tweet,
        "label": label,
        "compound": compound,
        "pos": scores['pos'],
        "neu": scores['neu'],
        "neg": scores['neg']
    })

print(f"Analyzed {len(results)} tweets.")
```

## Arrange Tweets by Sentiment Polarity

```python
arranged = sorted(results, key=lambda x: x['compound'], reverse=True)

print(f"{'Rank':<5} {'Label':<10} {'Compound':<10} {'Pos':<8} {'Neu':<8} {'Neg':<8} {'Tweet'}")
print("=" * 100)
for i, r in enumerate(arranged, 1):
    print(
        f"{i:<5} "
        f"{r['label']:<10} "
        f"{r['compound']:<10.4f} "
        f"{r['pos']:<8.4f} "
        f"{r['neu']:<8.4f} "
        f"{r['neg']:<8.4f} "
        f"{r['tweet'][:55]}..."
    )
```

## Sentiment Distribution Summary

```python
from collections import Counter

label_counts = Counter(r['label'] for r in results)
print("Sentiment Distribution:")
for label in ["Positive", "Neutral", "Negative"]:
    count = label_counts.get(label, 0)
    pct = count / len(results) * 100
    print(f"  {label:<10}: {count:>2} ({pct:.1f}%)")

avg_compound = sum(r['compound'] for r in results) / len(results)
print(f"\nAverage compound score: {avg_compound:.4f}")
if avg_compound >= 0.05:
    print("Overall corpus sentiment: Positive")
elif avg_compound <= -0.05:
    print("Overall corpus sentiment: Negative")
else:
    print("Overall corpus sentiment: Neutral")
```

## Detailed Per-Tweet Scores

```python
print(f"{'Tweet':<60} {'Label':<10} {'Compound':<10}")
print("=" * 85)
for r in results:
    truncated = r['tweet'][:57] + "..." if len(r['tweet']) > 57 else r['tweet']
    print(f"{truncated:<60} {r['label']:<10} {r['compound']:<10.4f}")
```
