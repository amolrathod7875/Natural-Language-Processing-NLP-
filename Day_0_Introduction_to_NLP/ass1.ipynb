# NLP Text Preprocessing with NLTK

This notebook demonstrates text preprocessing tasks including tokenization, stop-word removal, and stemming.

```python
import nltk
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
```

## 1. Sentence-Level Tokenization

Splitting a text into individual sentences.

```python
text = "Hello there. My name is Amol. I am learning Natural Language Processing."

sentences = nltk.sent_tokenize(text)

print("Original text:")
print(text)
print("\nSentence-level tokenization:")
for i, sentence in enumerate(sentences, 1):
    print(f"  Sentence {i}: {sentence}")
```

## 2. Word-Level Tokenization

Splitting a text into individual words (tokens).

```python
words = nltk.word_tokenize(text)

print("Word-level tokenization:")
print(words)
```

## 3. Stop-Word Removal

Removing common words (stop-words) that carry little semantic meaning.

```python
stop_words = set(stopwords.words('english'))

filtered_words = [word for word in words if word.lower() not in stop_words]

print("Original tokens:")
print(words)
print("\nStop-words:")
print(sorted(stop_words))
print("\nAfter stop-word removal:")
print(filtered_words)
```

## 4. Stemming

Reducing words to their root forms using the Porter Stemmer.

```python
stemmer = PorterStemmer()

sample_tokens = ['running', 'flies', 'organization', 'reading', 'better', 'connection', 'computing']

stemmed = [stemmer.stem(word) for word in sample_tokens]

print("Original words:")
print(sample_tokens)
print("\nStemmed words:")
print(stemmed)

print("\nStemming on filtered tokens:")
stemmed_filtered = [stemmer.stem(word) for word in filtered_words]
print(stemmed_filtered)
```
