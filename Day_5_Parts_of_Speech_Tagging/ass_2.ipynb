# NLP: POS Tagging and Lemmatization with NLTK

This notebook demonstrates Parts of Speech (POS) tagging and Lemmatization using NLTK libraries.

```python
import nltk
from nltk.tokenize import word_tokenize
from nltk import pos_tag
from nltk.stem import WordNetLemmatizer
from nltk.corpus import wordnet

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger')
nltk.download('averaged_perceptron_tagger_eng')
nltk.download('wordnet')
nltk.download('omw-1.4')
```

## 1. Parts of Speech (POS) Tagging

POS tagging assigns a part of speech (noun, verb, adjective, etc.) to each word in a sentence.

```python
text = "The quick brown fox jumps over the lazy dog"

tokens = nltk.word_tokenize(text)
tagged = nltk.pos_tag(tokens)

print("Tokens with POS tags:")
for word, tag in tagged:
    print(f"  {word:12} -> {tag}")
```

## 2. Understanding POS Tags

Common Penn Treebank POS tags used by NLTK:

```python
pos_tags = {
    'NN': 'Noun, singular',
    'NNS': 'Noun, plural',
    'VB': 'Verb, base form',
    'VBD': 'Verb, past tense',
    'VBG': 'Verb, gerund',
    'VBN': 'Verb, past participle',
    'VBP': 'Verb, present tense',
    'VBZ': 'Verb, 3rd person singular present',
    'JJ': 'Adjective',
    'JJR': 'Adjective, comparative',
    'JJS': 'Adjective, superlative',
    'RB': 'Adverb',
    'RBR': 'Adverb, comparative',
    'RBS': 'Adverb, superlative',
    'DT': 'Determiner',
    'IN': 'Preposition',
    'PRP': 'Personal pronoun',
    'CD': 'Cardinal number',
    'CC': 'Coordinating conjunction',
    'PDT': 'Predeterminer',
    'POS': 'Possessive ending'
}

print(f"{'Tag':<6} {'Description'}")
print('-' * 40)
for tag, desc in pos_tags.items():
    print(f"{tag:<6} {desc}")
```

## 3. POS Tagging on Multiple Sentences

```python
sentences = [
    "Natural language processing is fascinating.",
    "The students are reading interesting books.",
    "She quickly ran to the store."
]

for sentence in sentences:
    tokens = nltk.word_tokenize(sentence)
    tagged = nltk.pos_tag(tokens)
    print(f"Text: {sentence}")
    for word, tag in tagged:
        print(f"  {word:15} {tag}")
    print()
```

## 4. Mapping POS Tags to WordNet POS Tags

The WordNet lemmatizer requires WordNet POS tags (e.g., `wordnet.VERB`). This function converts Penn Treebank tags to WordNet tags.

```python
def get_wordnet_pos(treebank_tag):
    if treebank_tag.startswith('J'):
        return wordnet.ADJ
    elif treebank_tag.startswith('V'):
        return wordnet.VERB
    elif treebank_tag.startswith('N'):
        return wordnet.NOUN
    elif treebank_tag.startswith('R'):
        return wordnet.ADV
    else:
        return wordnet.NOUN

treebank_tag = 'VBG'
print(f"Penn tag '{treebank_tag}' -> WordNet POS: {get_wordnet_pos(treebank_tag)}")
treebank_tag = 'JJ'
print(f"Penn tag '{treebank_tag}' -> WordNet POS: {get_wordnet_pos(treebank_tag)}")
treebank_tag = 'NN'
print(f"Penn tag '{treebank_tag}' -> WordNet POS: {get_wordnet_pos(treebank_tag)}")
treebank_tag = 'RB'
print(f"Penn tag '{treebank_tag}' -> WordNet POS: {get_wordnet_pos(treebank_tag)}")
```

## 5. Lemmatization with POS Awareness

Lemmatization converts words to their dictionary base form (lemma). Unlike stemming, lemmatization produces valid words.

```python
lemmatizer = WordNetLemmatizer()

sample_words = [
    ('running', 'VBG'),
    ('geese', 'NNS'),
    ('better', 'JJR'),
    ('computing', 'VBG'),
    ('wolves', 'NNS'),
    ('easily', 'RB'),
    ('studies', 'NNS'),
]

print(f"{'Word':<15} {'Penn Tag':<10} {'Lemmatized (POS-aware)':<25} {'Lemma (NOUN only)'}")
print('-' * 80)

for word, tag in sample_words:
    wn_tag = get_wordnet_pos(tag)
    lemma_pos = lemmatizer.lemmatize(word, pos=wn_tag)
    lemma_noun = lemmatizer.lemmatize(word, pos=wordnet.NOUN)
    print(f"{word:<15} {tag:<10} {lemma_pos:<25} {lemma_noun:<20}")
```

## 6. Full Pipeline: Tokenize -> POS Tag -> Lemmatize

```python
text = "The geese are running quickly to the store."

tokens = nltk.word_tokenize(text)
tagged = nltk.pos_tag(tokens)

lemmatized = []
for word, tag in tagged:
    wn_tag = get_wordnet_pos(tag)
    lemma = lemmatizer.lemmatize(word, pos=wn_tag)
    lemmatized.append(lemma)

print(f"Original : {tokens}")
print(f"POS tags : {tagged}")
print(f"Lemmas   : {lemmatized}")
```

## 7. Comparing POS Tagging Results on a Rich Text

```python
rich_text = "NLTK provides easy-to-use interfaces to over 50 corpora and lexical resources."

tokens = nltk.word_tokenize(rich_text)
tagged = nltk.pos_tag(tokens)

print(f"{'Word':<20} {'POS Tag':<8} {'Lemma (POS-aware)'}")
print('-' * 55)

for word, tag in tagged:
    wn_tag = get_wordnet_pos(tag)
    lemma = lemmatizer.lemmatize(word, pos=wn_tag)
    print(f"{word:<20} {tag:<8} {lemma}")
```
