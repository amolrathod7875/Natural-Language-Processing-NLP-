```python
grammar = CFG.fromstring("""
S -> NP VP
NP -> Det N
VP -> V NP
Det -> "the" | "a"
N -> "dog" | "cat" | "rug"
V -> "chased" | "sat"
PUNCT -> "."
ADV -> "on"
S -> NP VP PUNCT
""")
```

```python
rd = RecursiveDescentParser(grammar)
```
